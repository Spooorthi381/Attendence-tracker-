```python
##### import datetime

def display_menu():
    print("\nAttendance Tracker")
    print("1. Add student")
    print("2. Mark attendance")
    print("3. Show attendance report for a date")
    print("4. Show summary for a date")
    print("5. Exit")

def add_student(students):
    name = input("Enter student name to add: ").strip()
    if name in students:
        print("Student already exists.")
    else:
        students.append(name)
        print(f"{name} added.")

def mark_attendance(attendance_records, students):
    date_str = input("Enter date (DD-MM-YY): ").strip()
    try:
        # Convert to date object
        date_obj = datetime.datetime.strptime(date_str, "%d-%m-%y").date()
    except ValueError:
        print("Invalid date format. Use DD-MM-YY.")
        return

    # Check if Sunday
    if date_obj.weekday() == 6:
        print(f"{date_str} is a Sunday. Marked as holiday for all students.")
        attendance_records[date_str] = {student: "Holiday" for student in students}
        return

    record = {}
    print(f"Marking attendance for {date_str}:")
    for student in students:
        while True:
            status = input(f"Enter status for {student} (P=Present, A=Absent): ").strip().upper()
            if status == 'P':
                record[student] = "Present"
                break
            elif status == 'A':
                record[student] = "Absent"
                break
            else:
                print("Invalid input. Enter 'P' or 'A'.")
    attendance_records[date_str] = record
    print(f"Attendance marked for {date_str}.")

def show_report(attendance_records, students):
    date_str = input("Enter date (DD-MM-YY): ").strip()
    if date_str not in attendance_records:
        print("No attendance record for this date.")
        return
    print(f"\nAttendance report for {date_str}:")
    for student in students:
        status = attendance_records[date_str].get(student, "Not Marked")
        print(f"{student}: {status}")

def show_summary(attendance_records, students):
    date_str = input("Enter date (DD-MM-YY): ").strip()
    if date_str not in attendance_records:
        print("No attendance record for this date.")
        return
    statuses = attendance_records[date_str].values()
    total_students = len(students)
    present = sum(1 for s in statuses if s == "Present")
    absent = sum(1 for s in statuses if s == "Absent")
    holidays = sum(1 for s in statuses if s == "Holiday")
    print(f"\nSummary for {date_str}:")
    print(f"Total students: {total_students}")
    print(f"Present: {present}")
    print(f"Absent: {absent}")
    print(f"Holiday: {holidays}")
    if holidays == total_students:
        print("It was a holiday for all students (probably Sunday).")

def main():
    students = []
    attendance_records = {}

    while True:
        display_menu()
        choice = input("Enter your choice: ")
        if choice == '1':
            add_student(students)
        elif choice == '2':
            if not students:
                print("No students in class. Add students first.")
            else:
                mark_attendance(attendance_records, students)
        elif choice == '3':
            show_report(attendance_records, students)
        elif choice == '4':
            show_summary(attendance_records, students)
        elif choice == '5':
            print("Exiting Attendance Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please select 1-5.")

if __name__ == "__main__":
    main()
```
